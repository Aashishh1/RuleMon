```python
# Numerical operations ke liye NumPy library import ki gayi hai
import numpy as np

# Data manipulation aur analysis ke liye pandas library import ki gayi hai
import pandas as pd

# Data visualization ke liye Matplotlib library ka pyplot module import kiya gaya hai
import matplotlib.pyplot as plt

# Advanced aur attractive visualizations ke liye Seaborn library import ki gayi hai
import seaborn as sns
```

```python
# Police data ko CSV file se pandas DataFrame mein load karna
# "/kaggle/input/police-data-code/3. Police Data.csv" file ka path diya gaya hai
police_df = pd.read_csv("3. Police Data.csv")
```

```python
# DataFrame ke har column mein missing (null) values ka count check karna
# Yeh step data cleaning ke liye important hota hai
police_df.isnull().sum()
```

```python
# 'country_name' column ko DataFrame se delete karna
police_df.drop('country_name', axis=1, inplace=True)

# 'search_type' column ko DataFrame se delete karna
police_df.drop('search_type', axis=1, inplace=True)
```

```python
# DataFrame ke saare column names ko list format mein display karna
police_df.columns
```

```python
# Saari rows ko drop karna jisme missing (null) values hain
police_df.dropna(inplace=True)

# Cleaned DataFrame ko display karna
police_df
```

```python
# 'stop_date' column ko datetime format mein convert karna
# Yeh step datetime-based operations aur filtering ke liye useful hota hai
police_df['stop_date'] = pd.to_datetime(police_df['stop_date'])
```

```python
# 'stop_time' column ko datetime format mein convert karna
# Ensure karna hoga ki 'stop_time' column mein valid time strings ho (e.g., "HH:MM:SS")
police_df['stop_time'] = pd.to_datetime(police_df['stop_time'])
```

```python
# 'stop_date' column se month extract karna aur naye 'month' column mein store karna
police_df['month'] = police_df['stop_date'].dt.month

# Naye 'month' column ko display karna
police_df['month']
```

```python
# 'stop_time' column se hour extract karna aur naye 'hour' column mein store karna
police_df['hour'] = police_df['stop_time'].dt.hour

# Naye 'hour' column ko display karna
police_df['hour']
```

```python
# Plot ke size ko define karna (width=8, height=6 inches)
plt.figure(figsize=(8, 6))

# Seaborn countplot ka use karke 'month' column ka count plot karna
# 'mako' palette se plot ka color scheme define kiya gaya hai
sns.countplot(x='month', data=police_df, palette='mako')

# X-axis aur Y-axis ke labels set karna
plt.xlabel('Occurance Month')  # X-axis label: Occurrence Month
plt.ylabel('Number Of Violations')  # Y-axis label: Number of Violations

# Plot ka title set karna
plt.title('Violation By Month')  # Title of the plot

# Plot ko better adjust karne ke liye tight layout lagana
plt.tight_layout()
```

```python
# Plot ke size ko define karna (width=8, height=6 inches)
plt.figure(figsize=(8, 6))

# Seaborn countplot ka use karke 'hour' column ka count plot karna
# 'mako' palette se plot ka color scheme define kiya gaya hai
sns.countplot(x='hour', data=police_df, palette='mako')

# X-axis aur Y-axis ke labels set karna
plt.xlabel('Occurance Hour')  # X-axis label: Occurrence Hour
plt.ylabel('Number Of Violations')  # Y-axis label: Number of Violations

# Plot ka title set karna
plt.title('Violation By Hour')  # Title of the plot

# Plot ko better adjust karne ke liye tight layout lagana
plt.tight_layout()
```

```python
# Plot ke size ko define karna (width=8, height=4 inches)
plt.figure(figsize=(8,4))

# Seaborn countplot ka use karke violation types ka count plot karna
# 'y' axis par 'violation' column hai (horizontal bar plot banega)
# 'hue' parameter se data ko 'driver_gender' ke basis par split kiya gaya hai
# 'order' parameter se violation types ko frequency ke descending order mein arrange kiya gaya hai
ax = sns.countplot(
    y='violation',
    data=police_df,
    palette='rocket',
    order=police_df.violation.value_counts().index,
    hue='driver_gender'
)

# X-axis aur Y-axis ke labels set karna
ax.set_xlabel('Number Of Occurrence', fontsize=10)  # X-axis label with font size
ax.set_ylabel('Type Of Violation')  # Y-axis label

# Plot ka title set karna
ax.set_title('Most Popular Violation Type', fontsize=15)  # Title with larger font size

# Legend (gender-wise distinction) ko display karna
ax.legend()

# Plot layout ko adjust karna for better spacing
plt.tight_layout()
```

```python
# Plot ke size ko define karna (width=8, height=4 inches)
plt.figure(figsize=(8,4))

# Seaborn countplot ka use karke 'stop_outcome' column ka horizontal bar plot banana
# 'order' parameter se outcomes ko descending frequency ke basis par arrange kiya gaya hai
sns.countplot(
    y='stop_outcome',
    data=police_df,
    palette='mako',
    order=police_df.stop_outcome.value_counts().index
)

# Y-axis ka label set karna
plt.ylabel('Stop Outcome', fontsize=12)  # Y-axis label with font size

# Plot ka title set karna
plt.title('Most Popular Stop Outcome', fontsize=15)  # Title with larger font size
```

```python
# 'is_arrested' column mein unique values (True/False) ka count karna
# Output descending order mein hoga (sabse zyada occurrence pehle)
arrested = police_df.value_counts('is_arrested')

# Result ko display karna
arrested
```

```python
# Plot ke size ko define karna (width=12, height=6 inches)
plt.figure(figsize=(12,6))

# Pie chart plot karna
# Parameters:
# - 'arrested': Values jo pie chart banane ke liye use hongi
# - 'labels': Labels for each segment (True/False from 'arrested.index')
# - 'labeldistance': Distance of labels from the center of the pie
# - 'autopct': Percentage format for each segment
# - 'startangle': Angle to start the pie chart from
# - 'wedgeprops': Edge properties for the wedges (e.g., color, line width)
# - 'colors': Custom colors for the pie chart
# - 'explode': First slice (True) ko slightly separate karna for emphasis
plt.pie(
    arrested,
    labels=arrested.index,
    labeldistance=1.0,
    autopct='%1.1f%%',
    startangle=0,
    wedgeprops={'linewidth': 3, 'edgecolor': 'white'},
    colors=['#F9E79F', '#F4D03F'],
    explode=(0.1, 0)
)

# Pie chart ka title set karna
plt.title('Percentage of Arrested Violators')

# Layout ko adjust karna for better spacing
plt.tight_layout()
```

```python
# DataFrame ko filter karna to include only rows where 'violation_raw' is 'Speeding'
# Phir 'driver_gender' column ke unique values (e.g., Male/Female) ka count karna
speed_violation = police_df[police_df['violation_raw'] == 'Speeding'].driver_gender.value_counts()

# Result ko display karna
speed_violation
```

```python
# Pie chart ke segments ke liye custom colors define karna
# '#088F8F': Teal color (e.g., for Male)
# '#FFB6C1': Light pink color (e.g., for Female)
colors = ['#088F8F', '#FFB6C1']

# Explode parameter define karna
# '0': Male segment ko center ke saath attached rakhna
# '0.1': Female segment ko slightly separate karna
explode = (0, 0.1)
```

```python
# Plot ke size ko define karna (width=12, height=6 inches)
plt.figure(figsize=(12,6))

# Pie chart plot karna
# Parameters:
# - 'speed_violation': Values jo pie chart banane ke liye use hongi (Male/Female counts)
# - 'labels': Labels for each segment (Male/Female from 'speed_violation.index')
# - 'explode': Explode parameter for separating specific segments
# - 'labeldistance': Distance of labels from the center of the pie
# - 'autopct': Format for displaying percentages on the pie chart
# - 'startangle': Starting angle for the pie chart
# - 'wedgeprops': Properties for the wedges (e.g., line width and edge color)
# - 'colors': Custom colors for each segment
plt.pie(
    speed_violation,
    labels=speed_violation.index,
    explode=explode,
    labeldistance=1.1,
    autopct='%1.1f%%',
    startangle=60,
    wedgeprops={'linewidth': 5, 'edgecolor': 'white'},
    colors=colors
)

# Pie chart ka title set karna
plt.title('Speed Violation Based On Gender', fontsize=15)

# Layout ko adjust karna for better spacing
plt.tight_layout()
```

```python
# 'driver_race' column mein unique values ka count karna
# Result descending order mein hoga (sabse zyada occurrence pehle)
race = police_df.value_counts('driver_race')

# Result ko display karna
race
```

```python
# Pie chart ke segments ke liye custom colors define karna
# '#D6EAF8': Light blue
# '#85929E': Greyish tone
# '#A9DFBF': Light green
# '#229954': Dark green
# '#21618C': Deep blue
colors1 = ['#D6EAF8', '#85929E', '#A9DFBF', '#229954', '#21618C']

# Explode parameter define karna
# Values:
# '0': No explosion (segment stays at the center)
# '0.2': Second segment will be slightly separated
# '0.1': Third segment will be slightly separated
# Other segments stay attached (value is 0)
explode1 = (0, 0.2, 0.1, 0, 0)
```

```python
# Plot ke size ko define karna (width=12, height=6 inches)
plt.figure(figsize=(12,6))

# Pie chart plot karna
# Parameters:
# - 'race': Values jo pie chart banane ke liye use hongi (Driver race counts)
# - 'labels': Labels for each segment (driver races from 'race.index')
# - 'labeldistance': Distance of labels from the center of the pie chart
# - 'autopct': Format for displaying percentages on the pie chart
# - 'startangle': Angle at which the pie chart will start
# - 'wedgeprops': Properties for the wedges (e.g., line width and edge color)
# - 'colors': Custom colors for each segment (defined in 'colors1')
# - 'explode': Explode parameter for separating specific segments (defined in 'explode1')
plt.pie(
    race,
    labels=race.index,
    labeldistance=1.2,
    autopct='%1.1f%%',
    startangle=210,
    wedgeprops={'linewidth': 1.7, 'edgecolor': 'white'},
    colors=colors1,
    explode=explode1
)

# Pie chart ka title set karna
plt.title('Violation Based On Race', fontsize=15)

# Layout ko adjust karna for better spacing
plt.tight_layout()
```

```python
# Seaborn's displot ka use karte hue 'driver_age' column ka distribution plot banana
# Parameters:
# - 'police_df['driver_age']': Data jo plot karna hai
# - 'kde=True': Kernel Density Estimate (smooth curve) ko include karna
# - 'bins=25': Distribution ke liye 25 bins define karna
# - 'color='#0E6655'': Plot ka color set karna (Dark green shade)
sns.displot(
    police_df['driver_age'],
    kde=True,
    bins=25,
    color='#0E6655'
)

# X-axis ka label set karna
plt.xlabel("Driver's Age", fontsize=10)  # Label with smaller font size

# Y-axis ka label set karna
plt.ylabel("Violations Occurrence", fontsize=10)  # Label with smaller font size

# Plot ka title set karna
plt.title('Violations Occurrence Based On Age', fontsize=15)  # Title with larger font size
```

```python
# Seaborn's displot ka use karte hue 'stop_duration' column ka distribution plot banana
# Parameters:
# - 'police_df['stop_duration']': Data jo plot karna hai
# - 'color='#0E6655'': Plot ka color set karna (Dark green shade)
sns.displot(
    police_df['stop_duration'],
    color='#0E6655'
)

# X-axis ka label set karna
plt.xlabel('Stop Duration', fontsize=10)  # X-axis label with smaller font size

# Plot ka title set karna
plt.title('Most Frequent Stop Duration', fontsize=15)  # Title with larger font size

# Layout ko adjust karna for better spacing
plt.tight_layout()
```

```python
# 'drugs_related_stop' column mein unique values ka count karna
# Result descending order mein return hoga
# True: Indicates a stop related to drugs
# False: Indicates a stop not related to drugs
drugs = police_df.value_counts('drugs_related_stop')

# Count ko display karna
drugs
```

```python
# Plot ke size ko define karna (width=12, height=6 inches)
plt.figure(figsize=(12,6))

# Pie chart plot karna
# Parameters:
# - 'drugs': Data jo pie chart banane ke liye use hoga (True/False counts for drug-related stops)
# - 'labels': Labels for each segment (True/False from 'drugs.index')
# - 'labeldistance': Labels ki distance from the center
# - 'autopct': Format for displaying percentages on the chart
# - 'startangle': Starting angle for the pie chart (0 degrees)
# - 'wedgeprops': Wedge properties (line width and edge color)
# - 'colors': Custom colors for each segment (Light green and Dark green)
plt.pie(
    drugs,
    labels=drugs.index,
    labeldistance=1.1,
    autopct='%1.1f%%',
    startangle=0,
    wedgeprops={'linewidth': 0.5, 'edgecolor': 'white'},
    colors=['#76D7C4', '#0E6655']
)

# Pie chart ka title set karna
plt.title('Drug Related Stops', fontsize=15)

# Layout ko adjust karna for better spacing
plt.tight_layout()
```
